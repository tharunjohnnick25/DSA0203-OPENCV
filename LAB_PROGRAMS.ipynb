```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read uploaded image
img = cv2.imread("image.jpg")

# Get image dimensions
rows, cols = img.shape[:2]

# Define source points
src_points = np.float32([
    [0, 0],
    [cols - 1, 0],
    [0, rows - 1],
    [cols - 1, rows - 1]
])

# Define destination points
dst_points = np.float32([
    [0, 0],
    [cols - 1, 0],
    [int(0.33 * cols), rows - 1],
    [int(0.66 * cols), rows - 1]
])

# Calculate perspective transformation matrix
M = cv2.getPerspectiveTransform(src_points, dst_points)

# Apply transformation
perspective_img = cv2.warpPerspective(img, M, (cols, rows))

# Display result
cv2_imshow(perspective_img)

# Save output
cv2.imwrite("Perspective_Transformed_Image.jpg", perspective_img)

print("Perspective transformation completed.")
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

def high_boost_filter(image, boost_factor):
    kernel_size = 3

    # Create averaging kernel
    kernel = np.ones(
        (kernel_size, kernel_size),
        dtype=np.float32
    ) / (kernel_size ** 2)

    # Blur image
    blur_image = cv2.filter2D(image, -1, kernel)

    # High boost filtering
    mask = image + (image - blur_image) * boost_factor

    # Convert values to valid pixel range
    mask = np.clip(mask, 0, 255)

    return np.uint8(mask)


# Read uploaded image
image_path = "image.jpg"
image = cv2.imread(image_path)

# Apply high boost filter
sharpened_image = high_boost_filter(image, 1.5)

# Display result
print("Original Image")
cv2_imshow(image)

print("High Boost Sharpened Image")
cv2_imshow(sharpened_image)

# Save output
cv2.imwrite("sharpened_image.jpg", sharpened_image)

print("High boost filtering completed.")
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read uploaded image in grayscale
image_path = "image.jpg"
image = cv2.imread(image_path, 0)

# Calculate Sobel gradients
gradient_x = cv2.Sobel(
    image,
    ddepth=cv2.CV_64F,
    dx=1,
    dy=0,
    ksize=3
)

gradient_y = cv2.Sobel(
    image,
    ddepth=cv2.CV_64F,
    dx=0,
    dy=1,
    ksize=3
)

# Combine gradients
gradient = cv2.subtract(gradient_x, gradient_y)

# Convert to 8-bit format
gradient = cv2.convertScaleAbs(gradient)

# Display result
cv2_imshow(gradient)

# Save output
cv2.imwrite("sharpened_image3.jpg", gradient)

print("Sobel gradient image saved successfully.")
```

```python
import cv2
from google.colab.patches import cv2_imshow

logo = cv2.imread("image.jpg")
img = cv2.imread("image.jpg")

# Check loading
if logo is None or img is None:
    print("Image not found")
else:
    h_logo, w_logo, _ = logo.shape
    h_img, w_img, _ = img.shape

    center_y = h_img // 2
    center_x = w_img // 2

    top_y = center_y - h_logo // 2
    left_x = center_x - w_logo // 2

    bottom_y = top_y + h_logo
    right_x = left_x + w_logo

    destination = img[top_y:bottom_y, left_x:right_x]

    result = cv2.addWeighted(destination, 1, logo, 0.5, 0)

    img[top_y:bottom_y, left_x:right_x] = result

    cv2_imshow(img)

    cv2.imwrite("watermarked_image.jpg", img)
```

```python
import cv2
from google.colab.patches import cv2_imshow

# Read uploaded video
cap = cv2.VideoCapture("video.mp4")

# Get total number of frames
num_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

# Read frames in reverse order
for i in reversed(range(num_frames)):
    cap.set(cv2.CAP_PROP_POS_FRAMES, i)

    ret, frame = cap.read()

    if not ret:
        continue

    # Display reverse frame
    cv2_imshow(frame)

cap.release()

print("Reverse video playback completed.")
```

```python
import cv2
from google.colab.patches import cv2_imshow

# Load Haar Cascade classifier
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

# Read uploaded video
cap = cv2.VideoCapture("video.mp4")

while True:
    ret, frame = cap.read()

    if not ret:
        break

    # Convert to grayscale
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Detect faces
    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(30, 30)
    )

    # Draw rectangles around faces
    for (x, y, w, h) in faces:
        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )

    # Display frame
    cv2_imshow(frame)

cap.release()

print("Face detection completed.")
```

```python
import cv2
from google.colab.patches import cv2_imshow

# Read uploaded image
img = cv2.imread("image.jpg")

# Define rectangle coordinates
start_point = (50, 50)
end_point = (200, 200)

# Rectangle properties
color = (0, 0, 255)   # Red color (BGR)
thickness = 2

# Draw rectangle
rect_img = cv2.rectangle(
    img.copy(),
    start_point,
    end_point,
    color,
    thickness
)

# Display image with rectangle
print("Image with Rectangle")
cv2_imshow(rect_img)

# Extract object (ROI)
obj_img = img[
    start_point[1]:end_point[1],
    start_point[0]:end_point[0]
]

# Display extracted object
print("Extracted Object")
cv2_imshow(obj_img)

# Save extracted object
cv2.imwrite("object.jpg", obj_img)

print("Object extracted and saved successfully.")
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read image
img = cv2.imread("image.jpg")

rows, cols = img.shape[:2]

# Four source points
src_points = np.float32([
    [0,0],
    [cols-1,0],
    [0,rows-1],
    [cols-1,rows-1]
])

# Destination points
dst_points = np.float32([
    [100,50],
    [cols-100,50],
    [50,rows-100],
    [cols-50,rows-100]
])

# Compute transformation matrix
M = cv2.getPerspectiveTransform(
    src_points,
    dst_points
)

# Apply transformation
transformed = cv2.warpPerspective(
    img,
    M,
    (cols,rows)
)

# Display
cv2_imshow(transformed)

cv2.imwrite(
    "DLT_Transformation.jpg",
    transformed
)
```

```python
import cv2
from google.colab.patches import cv2_imshow

# Read image in grayscale
img = cv2.imread(
    "image.jpg",
    0
)

# Apply Canny edge detection
edges = cv2.Canny(
    img,
    100,
    200
)

# Display output
cv2_imshow(edges)

# Save result
cv2.imwrite(
    "Canny_Edges.jpg",
    edges
)
```

```python
import cv2
from google.colab.patches import cv2_imshow

# Load image
img = cv2.imread("image.jpg")

# Load Haar cascade
watch_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_eye.xml"
)

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Detect object region
objects = watch_cascade.detectMultiScale(
    gray,
    scaleFactor=1.1,
    minNeighbors=5
)

# Draw rectangle
for (x,y,w,h) in objects:
    cv2.rectangle(
        img,
        (x,y),
        (x+w,y+h),
        (255,0,0),
        2
    )

# Display output
cv2_imshow(img)

cv2.imwrite("watch_detection.jpg", img)
```
